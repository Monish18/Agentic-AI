# Convert Text to Vector Embeddings Using HuggingFace in Python

```python
from langchain.embeddings import HuggingFaceEmbeddings

embedder = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

embedding = embedder.embed_query("My name is Monish!")
print(f"Vector length: {len(embedding)}")
print(f"First 5 dims: {embedding[:5]}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "What is the Capital of USA?",
    "Who is the Prime Minister of USA",
    "Who is Prime Minister of India"
    ]
document_embedding = embedder.embed_documents(docs)
```

```python
query = "Narendra Modi is the Prime Minister of India"
query_embedding = embedder.embed_query(query)
```

```python
cosine_similarity([query_embedding], document_embedding)
```
